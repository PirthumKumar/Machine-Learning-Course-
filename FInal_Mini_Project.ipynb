1. Data Cleaning 

```python
import pandas as pd

df = pd.read_csv("student_academic_performance_1000.csv")
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.info()
numeric_columns = ["Study_Hours_Per_Week" , "Attendance_Percent" , "Assignment_Score" , 
                    "Previous_GPA" ]

for col in numeric_columns:
    df[col].fillna(df[col].median() , inplace=True)
    
df[numeric_columns].isnull().sum()
```

```python
df.info()
```

```python
df["Extracurricular_Activities"].fillna("None" , inplace=True)
```

Visualizations & Explanation of Every Visulization

1. Scatter Plot
Attendance Percentage vs Final Score

Question Answered:
Does attendance influence students' final academic performance?

Purpose:
To analyze whether attendance is related to final academic performance.

Why this chart was selected:
A scatter plot is suitable for identifying relationships between two numerical variables.

What it shows:
The chart displays whether students with higher attendance percentages tend to achieve higher final scores.

Action/Decision Support:
Educational institutions can use this information to monitor attendance patterns and encourage consistent class participation.

```python
import matplotlib.pyplot as plt
import seaborn as sns


plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Attendance_Percent",
    y="Final_Score"
)

plt.title("Relationship Between Attendance and Final Score")
plt.xlabel("Attendance Percentage")
plt.ylabel("Final Score")
plt.show()
```

2.Line Chart
Study Hours vs Average Final Score

Question Answered:
Do increased study hours improve academic performance?

Purpose:
To understand how study time affects academic performance.

Why this chart was selected:
A line chart clearly shows changes and trends across different study-hour levels.

What it shows:
The chart demonstrates how average final scores change as students spend more time studying.

Action/Decision Support:
Helps educators recommend effective study routines for students.

```python
# Calculate average score by study hours

study_score = (
    df.groupby("Study_Hours_Per_Week")
    ["Final_Score"]
    .mean()
    .reset_index()
)


plt.figure(figsize=(9,5))
sns.lineplot(
    data=study_score,
    x="Study_Hours_Per_Week",
    y="Final_Score",
    marker="o"
)


plt.title(
    "Average Final Score Based on Study Hours"
)
plt.xlabel(
    "Study Hours Per Week"
)
plt.ylabel(
    "Average Final Score"
)


plt.show()
```

3.Correlation Heatmap
Academic Factors Correlation

Question Answered:
Which academic factors have the strongest relationship with final performance?

Purpose:
To identify which academic factors have the strongest relationship with final scores.

Why this chart was selected:
A heatmap allows multiple numerical relationships to be viewed together.

What it shows:
It highlights positive and negative relationships between academic variables.

Action/Decision Support:
Helps identify the most important academic factors affecting final results.

```python
academic_cols = [
    "Study_Hours_Per_Week",
    "Attendance_Percent",
    "Assignment_Score",
    "Midterm_Score",
    "Final_Score",
    "Previous_GPA"

]
corr = df[academic_cols].corr()
plt.figure(figsize=(9,7))
sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)
plt.title(
    "Correlation Between Academic Variables"
)
plt.show()
```

4. Box Plot
Assessment Score Distribution
Question Answered:

How do assignment and midterm scores vary among students?

Purpose:

To compare assessment score variation among students.

Why this chart was selected:

Box plots show:

Median values
Score variation
Spread
Possible outliers
What it shows:

The distribution and differences in assessment performance.

Action/Decision Support:
Helps identify inconsistent assessment performance and students who may need additional support.

```python
# Convert scores into long format

score_data = df[
    [
        "Assignment_Score",
        "Midterm_Score",
        "Final_Score"
    ]
]


plt.figure(figsize=(8,5))


sns.boxplot(
    data=score_data
)


plt.title(
    "Distribution of Academic Assessment Scores"
)


plt.ylabel(
    "Score"
)


plt.show()
```

5. Bubble Chart

Attendance, Final Score and Study Performance
Question Answered:

How do attendance and study habits together influence final performance?

Purpose:

To analyze three factors together:

Attendance
Final Score
Study Hours

Why this chart was selected:

A bubble chart displays three variables simultaneously.

What it shows:

Students with higher study hours appear as larger bubbles, allowing comparison of study effort and academic outcomes.

Action/Decision Support:

Helps identify whether students with lower scores require improved study strategies.

```python
plt.figure(figsize=(9,6))


plt.scatter(

    df["Attendance_Percent"],

    df["Final_Score"],

    s=df["Study_Hours_Per_Week"]*20,

    alpha=0.5

)


plt.xlabel(
    "Attendance Percentage"
)


plt.ylabel(
    "Final Score"
)


plt.title(
    "Attendance vs Final Score (Bubble Size = Study Hours)"
)


plt.show()
```

Findings 

Finding 1: 
Attendance shows a relationship with final scores. Students who maintain higher attendance levels and perform well generally achieve better final scores.

Finding 2 :
Assignment scores, midterm performance, attendance percentage, and academic performance are strongly connected with final scores. Students who perform well in continuous assessments tend to achieve higher final scores.

Finding 3 :
Study habits, such as study hours per week, consistency, and regular effort, influence academic outcomes. Students who give importance to their studies and dedicate more hours generally demonstrate better academic performance, although study time alone does not determine success.

-Conclusion

The findings indicate that academic success is influenced by multiple factors. Regular attendance, effective study habits, and strong performance in assignments and midterm examinations contribute to improved final results.
